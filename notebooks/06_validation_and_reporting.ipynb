# 06 — Final Validation and Reporting

**Purpose.** Re-evaluate every configuration with Sionna-RT on **held-out
scenarios** and compare baseline, TuRBO and MARL on identical terms. PROJECT.md
section 16, Phases 7 and 8, plus sections 17, 19 and 20.

### Test-scenario discipline

The split is at **scenario** level and has been frozen since notebook 03
(PROJECT.md section 12.3, Decision 8). The optimized configurations were found
against training scenarios; they are validated here on scenarios the surrogate
never saw — different buildings, different materials, different UE trajectories.

That is the sim-to-reality question of section 12, and it is the whole point of
this notebook. A configuration that wins on the scenario it was tuned for and
loses on a perturbed one has told you something important, and only a
scenario-level boundary can reveal it.

UE density is rebuilt per test scenario. The Band Priority Score weights
locations by where users are, so scoring the optimized network on the same
mobility realisation that shaped its objective is exactly the leakage the split
exists to prevent.

**Every number in this notebook comes from Sionna-RT.** A KPI improvement that
exists only in the surrogate is not a result (docs/adr/0003).

## 2. Load the three configurations

Baseline, TuRBO, MARL. The baseline is not optional: every reported improvement
is stated relative to the current network, and a baseline evaluated in a
different run — against a different scene build or a different grid — is not a
valid reference.

All three must satisfy `tilt_min <= tilt <= tilt_max` for every cell-band. An
infeasible final configuration is not a result at any KPI value (PROJECT.md
section 25.2), so it is asserted here rather than assumed.

```python
import json

from src.data.load import load_cell_config
from src.optim.space import TiltSpace
from src.radio import cell_band

table = cell_band.build_table(load_cell_config(cfg), cfg)
space = TiltSpace(table, cfg)

from src.radio import sampling

results_dir = Path("reports/results")
tilts = {"baseline": space.baseline()}
for method in ("bo", "marl"):
    path = results_dir / f"{method}_tilt.json"
    tilts[method] = np.array(json.loads(path.read_text())["tilt"])

for method, tilt in tilts.items():
    sampling.assert_within_bounds(tilt, table)  # PROJECT.md section 24, constraint validation
print("all configurations feasible")

pd.DataFrame(tilts, index=table["gcell_id"] + "|" + table["band"]).head(10)
```

## 4. KPI comparison — PROJECT.md section 20

In the priority order from `cfg.kpi.order`, not alphabetically. The order *is*
the objective, and a table that hides it invites the reader to weigh weak rate
equally with hole rate. That order is now hole, overlap, mean overlap
neighbours, Band Priority Score, weak — weak rate last.

All five are reported separately. A single aggregate score is explicitly a
failure mode (PROJECT.md section 25.7): the five have different operational
meanings and a strict priority between them.

The delta column is signed so that positive always means better — four KPIs
improve by decreasing and one, BPS, by increasing, so an unlabelled delta gets
misread exactly once per reader (section 20).

```python
from src.evaluation import compare

kpi_table = compare.kpi_table(results, cfg)
kpi_table
```

Mean, standard deviation, **best and worst**. A method whose worst seed opens a

## 7. Tilt tables — PROJECT.md section 19

The deliverable a network engineer actually receives: one row per cell-band with
current tilt, optimized tilt and `delta_tilt = optimized_tilt - current_tilt`.

`delta_tilt` is derived here purely to communicate how far each antenna moves.
It was never optimised and carries no penalty (PROJECT.md Decision 1 and
section 25.1), and every `delta_tilt` in the report must be computed from the
same current-tilt baseline (section 24).

Sorted by absolute offset: the rows that matter are the antennas that move most,
and the cell identifiers are opaque hashes that sort into no useful order.

```python
from src.evaluation import report

tilt_tables = {m: report.tilt_table(tilts[m], table) for m in ("bo", "marl")}
tilt_tables["bo"].head(15)
```

## 8. Spatial maps — PROJECT.md section 16 Phase 8

Where the numbers become an argument. The KPI table says hole rate fell three
points; these say whether the remaining holes moved to the edge of the area or
opened up where the users are.

The difference maps are the check for a bad trade: a configuration that improves
the aggregate by degrading a dense area is visible here and nowhere in section 4.

```python
from src.evaluation import analysis
from src.kpi import serving

for method in ("baseline", "bo", "marl"):
    analysis.rsrp_map(results[method]["rsrp"], cfg, title=f"{method} RSRP")
    analysis.coverage_map(results[method]["rsrp"], cfg)

analysis.difference_map(
    serving.max_rsrp(results["baseline"]["rsrp"]),
    serving.max_rsrp(results["bo"]["rsrp"]),
    cfg,
)
analysis.dominant_band_map(results["bo"]["dominant_band"], table, cfg)
```

## 9. Surrogate versus ground truth

How far the surrogate was off at the configurations that mattered. This closes
the loop on notebook 04: error was measured there on sampled configurations, and
this is the error at the optimum each method actually selected — the only place
it can be measured, and the place an optimizer's search for the surrogate's
optimistic regions would show up.

Report it at both levels, as in notebook 04: radio-map error in dB, and the KPI
error that produces. A gap larger than the improvement being claimed means the
result is not supported, however good the ground-truth number looks alone.

TuRBO and MARL do not lean on the surrogate equally — MARL far harder — so part
of any difference between them is a difference in how much approximation each
tolerated. State that; do not explain it away (docs/adr/0003).

```python
gaps = pd.DataFrame({m: results[m]["gap"] for m in ("bo", "marl")})
gaps["claimed_improvement_bo"] = kpi_table["bo_delta"]
gaps.loc[list(cfg.kpi.order)]
```
